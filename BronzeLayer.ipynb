### **Data Reading**

**variables-widgets**

```python
dbutils.widgets.text("file_name","")
```

```python
P_file_name = dbutils.widgets.get("file_name")
```

**Streaming-exactlyOnce processing**

```python
df = spark.readStream.format("cloudFiles")\
            .option("cloudFiles.format","parquet")\
            .option("cloudFiles.schemaLocation",f"abfss://bronzestg@storageloweretep1.dfs.core.windows.net/checkpoint_{P_file_name}")\
            .load(f"abfss://btsdatamart@storageloweretep1.dfs.core.windows.net/{P_file_name}")
```

**Data Writing**

```python
df.writeStream.format("parquet")\
    .outputMode("append")\
    .option("checkpointLocation",f"abfss://bronzestg@storageloweretep1.dfs.core.windows.net/checkpoint_{P_file_name}")\
    .option("path",f"abfss://bronzestg@storageloweretep1.dfs.core.windows.net/{P_file_name}")\
    .trigger(once=True)\
    .start()
```
