### **Data Reading**

**Streaming-exactlyOnce**

```python
df = spark.readStream.format("cloudFiles")\
            .option("cloudFiles.format","parquet")\
            .option("cloudFiles.schemaLocation","abfss://bronzestg@storageloweretep1.dfs.core.windows.net/checkpoint_orders")\
            .load("abfss://btsdatamart@storageloweretep1.dfs.core.windows.net/orders")
```

```python
df.writeStream.format("parquet")\
    .outputMode("append")\
    .option("checkpointLocation","abfss://bronzestg@storageloweretep1.dfs.core.windows.net/checkpoint_orders")\
    .option("path","abfss://bronzestg@storageloweretep1.dfs.core.windows.net/orders")\
    .trigger(once=True)\
    .start()
```
